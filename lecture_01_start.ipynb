```python
# Create variables of different types
n = 1
x = -0.5
s = "Hello, World!"
print(n, x, s)
```

```python
# Create a list of mixed types
mixed_list = [n, x, s, [1, 2, 3], {"key": "value"}]
print(mixed_list)
```

```python
# Calculate the length of the mixed_list
N = len(mixed_list)
print(f"Length of mixed_list: {N}")
# NB: f-strings are used for insrting variables into strings
```

```python
# Create a for loop
for i in range(N):
    print(f"Iteration {i}: {mixed_list[i]}")
```

```python
# Create a for loop with a conditional statement
for i in range(N):
    if isinstance(mixed_list[i], str):
        print(f"String at index {i}: {mixed_list[i]}")
    else:
        print(f"Iteration {i}: the value {mixed_list[i]} is not a string")
```
