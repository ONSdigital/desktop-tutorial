# Motivation
Functions in Python, are used so that you don't have to repeat snippets of code. Functions are building blocks or larger programs. The body of the function does not change if you run it in different places, but the inputs (arguments) and outputs usually do.

#example


````python
def incremement(x):
    return x + 1


incremement(2)
````

*NB* Definition of a function starts with def
After the function name there must be round parentethes and colon sign
Body must be indented with respect to the def, 4 spaces is default.
Function ends with the word return. Sometimes, return occurs inside the body:

````python
def absolute_value(x):
    if x >= 0:
        return x
    else:
        return -x
````

````python
print(absolute_value(1))
print(absolute_value(0))
print(absolute_value(-10))
````

*NB* Functions allow us avoiding code repetition
Example: calculate binomial coefficients
Use factorial function as a helper

$$C_n^k = \frac{n!}{k!(n-k)!}$$

````python
def factorial(n):
    """Calculate factorial of n."""

    f = 1
    for i in range(n):
        f *= i + 1
    return f
````

````python
factorial(5)
````

````python
def binomial(n, k):
    return factorial(n) / (factorial(k) * factorial(n - k))


binomial(4, 2)
````

## Positional argumens
The order matters!

````python
def power(n, p):
    return n**p


print(power(2, 3))
print(power(3, 2))
````

*NB* The number of positional, non-default arguments in the function ddefinition and its call must match.

````python
# power(1, 2, 3)
````

## Keyword arguments
Sometimes, arguments are changed rarely, and you don't  want to specify them every time. 
Keyword arguments can give provide some default values. You may not use them, or you may choose 
using them to give a different value.

````python
def increment(n, step=1):
    return n + step


print(increment(10))
print(increment(10, 100))
print(increment(10, step=3))
````

*NB* In the function definition, all keyword arguments must follow the non-default arguments

````python
# def bad_increment(step=1, n):
#     return n + step
````

*NB* Usually, it's a good style to put spaces before anf after =, such as x = 0.
For keyword arguments, there should be no spaces: increment(10, step=3)

## Outputs
### Multiple outputs - Tuple
A function my return multiple variables, separated by a comma.

````python
def quadratic(a, b, c):
    """Calculate real roots of a quadratic equation"""
    d = b**2 - 4 * a * c
    if d < 0:
        return None
    elif d == 0:
        x = -b / (2 * a)
        return x
    else:
        x1 = (-b - d**0.5) / (2 * a)
        x2 = (-b + d**0.5) / (2 * a)
        return x1, x2
````

````python
print(quadratic(1, 2, 2))
print(quadratic(1, 2, 1))
print(quadratic(1, 2, 0))

print(type(quadratic(1, 2, 0)))
````
