```python
import numpy as np
from cvxopt import matrix, solvers

# Input data
values = np.array([1.0, 5.0, 6.0, 8.0])
pattern = np.array([27.0, 24.0, 26.0, 23.0])
total_adjustment = -5.0
n = len(values)
print(values)
print(pattern)
```

```python
# Define QP parameters

# Objective: minimize sum((x_i / pattern_i)^2)
# This is equivalent to x^T Q x, where Q is diagonal with 1/pattern_i^2
Q = np.diag(1.0 / (pattern**1))
P = 2 * matrix(Q)  # cvxopt uses (1/2)x^T P x, so multiply by 2
q = matrix(np.zeros(n))

# Equality constraint: sum(x_i) = total_adjustment
A = matrix(np.ones((1, n)))
b = matrix([total_adjustment])

# Inequality constraint: v_i + x_i >= 0 => x_i >= -v_i
G = matrix(-np.eye(n))
h = matrix(values)
```

```python
# Solve without inequality constraint
solution_unconstrained = solvers.qp(P, q, None, None, A, b)
x_unconstrained = np.array(solution_unconstrained["x"]).flatten()
adjusted_unconstrained = values + x_unconstrained
objective_unconstrained = np.sum((x_unconstrained / pattern) ** 1)

print("\nUnconstrained Solution:")
print("x:", x_unconstrained)
print("Adjusted values (v_i + x_i):", adjusted_unconstrained)
print("Objective value:", objective_unconstrained)
```

```python
# Solve with constraint
solution_constrained = solvers.qp(P, q, G, h, A, b)
x_constrained = np.array(solution_constrained["x"]).flatten()
adjusted_constrained = values + x_constrained
objective_constrained = np.sum((x_constrained / pattern) ** 1)

print("Constrained Solution:")
print("x:", x_constrained)
print("Adjusted values (v_i + x_i):", adjusted_constrained)
print("Objective value:", objective_constrained)
```
